### Aufgabe 4 – Break-Even-Analyse

Im Rahmen einer Unternehmensgründung soll eine Funktion zur Break-Even-Analyse implementiert werden. Der Break-Even-Punkt (BEP) gibt an, ab welcher Absatzmenge ein Unternehmen seine Gesamtkosten durch Erlöse deckt:

$$BEP = \frac{\text{Fixkosten}}{\text{Preis} - \text{variable Stückkosten}}$$

**a)** Schreiben Sie eine Funktion `break_even(fixkosten, preis, variable_kosten)`, die den BEP (gerundet auf ganze Stück) zurückgibt. Geben Sie eine Fehlermeldung aus und geben Sie `None` zurück, falls `preis <= variable_kosten` (kein positiver Deckungsbeitrag).

**b)** Rufen Sie die Funktion mit folgenden Szenarien auf und geben Sie das Ergebnis jeweils als aussagekräftigen Satz aus:
- Fixkosten: 50.000 €, Preis: 25 €, variable Kosten: 10 €
- Fixkosten: 12.000 €, Preis: 8 €, variable Kosten: 8 € *(Fehlerfall)*
- Fixkosten: 80.000 €, Preis: 120 €, variable Kosten: 45 €

<details>
<summary><b>Tipp zu a) – Formel umsetzen</b></summary>

```python
import math

def break_even(fixkosten, preis, variable_kosten):
    if preis <= variable_kosten:
        print("Fehler: Kein positiver Deckungsbeitrag.")
        return None
    bep = fixkosten / (preis - variable_kosten)
    return math.ceil(bep)  # aufrunden auf ganze Stück
```
</details>

<details>
<summary><b>Tipp zu b) – None prüfen</b></summary>

Fangen Sie den Fehlerfall ab, bevor Sie das Ergebnis ausgeben:
```python
bep = break_even(...)
if bep is not None:
    print(f"Break-Even wird bei {bep} Stück erreicht.")
```
</details>

````python
Kennzahlen = [(50000, 25, 10), (12000, 8, 8), (80000, 120, 45)]
def break_even(fixkosten, preis, variable_kosten):
    if preis <= variable_kosten:
        return None
    bep = fixkosten/(preis-variable_kosten)
    return bep

for fixkosten, preis, variable_kosten in Kennzahlen:
    bep = break_even(fixkosten, preis, variable_kosten)
    if bep is not None:
        print(f'Break-Even wird bei {int(bep)} Stück erreicht.')
````

- `display()` – gibt eine Übersicht des Kurses aus (Titel, Dozent:in, aktuelle Belegung)

**a)** Implementieren Sie die Klasse und führen Sie anschließend den Testcode aus.

    if len(self.teilnehmer) >= self.max_teilnehmer:

<details>
<summary><b>Tipp zu b) – private Attribute</b></summary>

Ersetzen Sie `self.titel` durch `self.__titel` usw. Innerhalb der Klasse verwenden Sie dann ebenfalls `self.__titel`.
</details>

````python
# Aufgabe 5a – Klasse implementieren
class Kurs:
    def __init__(self, titel, dozent, max_teilnehmer):
        self._titel = titel
        self._dozent = dozent
        self._max_teilnehmer = max_teilnehmer
        self._teilnehmer = []
    def anmelden(self, name):
        self.name = name
        if len(self._teilnehmer) < self._max_teilnehmer:
            self._teilnehmer.append(self.name)
            print(f'{self.name} wurde angemeldet.')
        else:
            print(f'Anmeldung von {self.name} fehlgeschlagen: Kurs ist voll.')
    def abmelden(self, name):
        if self.name in self._teilnehmer:
            self._teilnehmer.remove(self.name)
            print(f'{self.name} wurde abgemeldet.')
        else:
            print(f'Abmeldung von {self.name} fehlgeschlagen: Teilnehmer nicht gefunden.')
    def display(self):
        print(f'Der Kurs "{self._titel}" von {self._dozent} hat {len(self._teilnehmer)} Teilnehmer:')
    

# Testcode (erst ausführen, wenn die Klasse fertig ist):
bwl_grundlagen = Kurs("BWL Grundlagen", "Prof. Dr. Richter", 3)
bwl_grundlagen.anmelden("Anna Bauer")
bwl_grundlagen.anmelden("Ben Schulz")
bwl_grundlagen.anmelden("Clara Hoffmann")
bwl_grundlagen.anmelden("Daniel Meier")  # -> Fehlermeldung erwartet
bwl_grundlagen.abmelden("Ben Schulz")
bwl_grundlagen.abmelden("Felix Wagner")  # -> Fehlermeldung erwartet
bwl_grundlagen.display()
````

````python
# Aufgabe 6a – Klasse OnlineKurs implementieren
class OnlineKurs(Kurs):
    def __init__(self, titel, dozent, max_teilnehmer, plattform):
        super().__init__(titel, dozent, max_teilnehmer)
        self.plattform = plattform
    
    def get_link(self):
        titel_link = self._titel.replace(" ", "_").lower()
        return f"https:// {self.plattform}.com/{titel_link}"
    
    def display(self):
        super().display()
        print(f'Plattform: {self.plattform}')
        print(f'Link: {self.get_link()}')

# Testcode:
stats_online = OnlineKurs("Statistik I", "Prof. Dr. Lange", 100, "Zoom")
stats_online.anmelden("Anna Bauer")
stats_online.anmelden("Ben Schulz")
stats_online.display()
````
